```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Indian_Insurance_Data.csv')
```

```python
def age_group(age):
    if age < 18:
        return 'Junior'
    elif age < 45:
        return 'Middle Age'
    else:
        return 'Senior'
```

```python
df['age_group'] = df['age'].apply(age_group)
```

```python
df.head()
```

```python
df['height'] = df['height']/100
```

```python
df['bmi'] = df['weight']/(df['height'])**2
```

```python
df.head()
```

```python
def bmi_category(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif bmi < 24.9:
        return 'Normal'
    elif bmi < 29.9:
        return 'Overweight'
    else:
        return 'Obese'
```

```python
df['bmi_category'] = df['bmi'].apply(bmi_category)
```

```python
df.sample(5)
```

```python
df['smoker'].dtype
```

```python
def lifestyle_risk(row):

    bmi = row['bmi_category']
    smoker = row['smoker']

    if bmi == 'Normal' and smoker == False:
        return 'low'

    elif bmi == 'Underweight' and smoker == False:
        return 'medium'

    elif bmi == 'Overweight' and smoker == False:
        return 'medium'

    elif bmi == 'Normal' and smoker == True:
        return 'medium'

    elif bmi == 'Obese' and smoker == False:
        return 'high'

    elif bmi == 'Underweight' and smoker == True:
        return 'high'

    elif bmi == 'Overweight' and smoker == True:
        return 'high'

    elif bmi == 'Obese' and smoker == True:
        return 'high'


df['lifestyle_risk'] = df.apply(lifestyle_risk, axis=1)
```

```python
df.sample(5)
```

```python
df['city'].unique()
```

```python
def city_tier(city):

    tier_1 = [
        'Mumbai',
        'Delhi',
        'Bangalore',
        'Hyderabad',
        'Chennai',
        'Kolkata',
        'Pune',
        'Ahmedabad',
        'Ghaziabad',
        'Faridabad'
    ]

    tier_2 = [
        'Jaipur',
        'Lucknow',
        'Kanpur',
        'Nagpur',
        'Indore',
        'Bhopal',
        'Patna',
        'Vadodara',
        'Surat',
        'Nashik',
        'Visakhapatnam',
        'Ranchi',
        'Ludhiana',
        'Agra',
        'Varanasi',
        'Rajkot',
        'Meerut',
        'Amritsar'
    ]

    tier_3 = [
        'Allahabad',
        'Srinagar'
    ]

    if city in tier_1:
        return '1'

    elif city in tier_2:
        return '2'

    elif city in tier_3:
        return '3'

    else:
        return 'Unknown'
```

```python
df['city_tier'] = df['city'].apply(city_tier)
```
